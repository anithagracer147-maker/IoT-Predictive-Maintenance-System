```python
temperature = [35, 42, 55, 65, 75, 85]
vibration = [20, 30, 45, 60, 70, 90]

print("Temperature:", temperature)
print("Vibration:", vibration)
```

```python
for i in range(len(temperature)):
    if temperature[i] >= 80 or vibration[i] >= 80:
        status = "CRITICAL"
    elif temperature[i] >= 60 or vibration[i] >= 60:
        status = "WARNING"
    else:
        status = "NORMAL"

    print(temperature[i], "C |",
          vibration[i], "% |", status)
```

```python
import matplotlib.pyplot as plt

plt.plot(temperature, marker='o')
plt.title("Motor Temperature Monitoring")
plt.xlabel("Reading Number")
plt.ylabel("Temperature (°C)")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.plot(vibration, marker='o')
plt.title("Motor Vibration Monitoring")
plt.xlabel("Reading Number")
plt.ylabel("Vibration (%)")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.plot(temperature, marker='o', label='Temperature (°C)')
plt.plot(vibration, marker='s', label='Vibration (%)')

plt.title("Motor Condition Monitoring")
plt.xlabel("Reading Number")
plt.ylabel("Value")
plt.legend()
plt.grid(True)
plt.show()
```

```python
normal = 0
warning = 0
critical = 0

for i in range(len(temperature)):
    if temperature[i] >= 80 or vibration[i] >= 80:
        critical += 1
    elif temperature[i] >= 60 or vibration[i] >= 60:
        warning += 1
    else:
        normal += 1

print("Motor Condition Summary")
print("-----------------------")
print("Normal readings:", normal)
print("Warning readings:", warning)
print("Critical readings:", critical)
print("Total readings:", len(temperature))
```
